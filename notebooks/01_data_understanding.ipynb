## 3. Análise de Valores Ausentes

As pesquisas possuem grande quantidade de variáveis e apresentam perguntas condicionais, nas quais determinadas questões são respondidas apenas por grupos específicos de participantes.

Por esse motivo, a presença de valores ausentes deve ser analisada antes de qualquer estratégia de tratamento, evitando interpretar automaticamente valores nulos como problemas de qualidade dos dados.

```python
resumo_nulos = pd.DataFrame({
    "Base": ["2023-2024", "2024-2025", "2025-2026"],

    "Total de células": [
        df_2023.size,
        df_2024.size,
        df_2025.size
    ],

    "Células nulas": [
        df_2023.isna().sum().sum(),
        df_2024.isna().sum().sum(),
        df_2025.isna().sum().sum()
    ]
})

resumo_nulos["Percentual de nulos (%)"] = (
    resumo_nulos["Células nulas"] /
    resumo_nulos["Total de células"] * 100
).round(2)

resumo_nulos
```

### Interpretação dos Valores Ausentes

As três pesquisas apresentam elevada proporção de valores ausentes, correspondendo a **56,68%** das células em 2023-2024, **57,20%** em 2024-2025 e **61,04%** em 2025-2026.

Entretanto, esses percentuais não devem ser interpretados isoladamente como problemas de qualidade dos dados. O questionário contém perguntas condicionais e grupos de variáveis aplicáveis apenas a determinados perfis ou respostas anteriores, fazendo com que a ausência de informação possa representar um comportamento esperado da pesquisa.

Observa-se também aumento do percentual global de valores ausentes na edição mais recente. Antes da definição de qualquer estratégia de tratamento, será necessário avaliar os valores nulos no contexto de cada variável e de sua função no questionário.

Dessa forma, não será aplicada nesta etapa uma estratégia generalizada de preenchimento ou remoção de valores ausentes. O tratamento será definido posteriormente de acordo com as variáveis selecionadas para as análises e com as regras de negócio associadas a cada indicador.

### 3.1 Variáveis com Maior Percentual de Valores Ausentes

Para compreender melhor a distribuição dos valores ausentes, são identificadas as variáveis com maior percentual de dados nulos em cada edição da pesquisa.

```python
def top_colunas_nulas(df, n=10):
    percentual = (df.isna().mean() * 100).round(2)

    return (
        percentual
        .sort_values(ascending=False)
        .head(n)
        .reset_index()
        .rename(columns={
            "index": "Variável",
            0: "Percentual de nulos (%)"
        })
    )

nulos_2023 = top_colunas_nulas(df_2023)
nulos_2024 = top_colunas_nulas(df_2024)
nulos_2025 = top_colunas_nulas(df_2025)
```

```python
print("2023-2024")
display(nulos_2023)

print("\n2024-2025")
display(nulos_2024)

print("\n2025-2026")
display(nulos_2025)
```

### Interpretação das Variáveis com Maior Percentual de Nulos

A análise das variáveis com maior percentual de valores ausentes reforça que a ausência de dados está concentrada, em grande parte, em perguntas específicas ou condicionais do questionário.

Nas edições de 2024-2025 e 2025-2026, algumas das maiores taxas de ausência aparecem em variáveis relacionadas à busca por oportunidades profissionais, experiência em processos seletivos e determinadas tecnologias ou ambientes de trabalho.

Também é possível observar padrões semelhantes de ausência entre as duas edições mais recentes, o que sugere continuidade na estrutura condicional de determinadas perguntas.

Portanto, elevados percentuais de valores nulos não serão tratados automaticamente como problemas de qualidade. A estratégia de tratamento deverá considerar o contexto de cada variável e sua utilização nas análises, evitando remoções ou imputações indiscriminadas.

### 4.1 Eixos Analíticos

Os principais eixos inicialmente considerados são:

- **Perfil demográfico:** idade, gênero e raça/etnia;
- **Localização:** estado, região e país;
- **Perfil profissional:** cargo, área de atuação, experiência e senioridade;

- **Modelo de trabalho:** presencial, híbrido e remoto;

- **Inteligência Artificial:** adoção, utilização e impactos de IA/IA Generativa;
- **Mercado de trabalho:** oportunidades, empregabilidade e características profissionais.

A seleção definitiva das variáveis será realizada considerando a disponibilidade e a equivalência semântica entre as três edições da pesquisa.

```python
termos_interesse = [
    "genero",
    "idade",
    "raca",
    "etnia",
    "estado",
    "regiao",
    "cargo",
    "salario",
    "senior",
    "experiencia",
    "trabalho",
    "remoto",
    "hibrido",
    "tecnologia",
    "linguagem",
    "cloud",
    "inteligencia artificial",
    "generativa",
    "llm"
]

def buscar_variaveis(df, termos):
    resultados = []

    for coluna in df.columns:
        nome = str(coluna).lower()

        if any(termo in nome for termo in termos):
            resultados.append(coluna)

    return resultados

variaveis_2023 = buscar_variaveis(df_2023, termos_interesse)
variaveis_2024 = buscar_variaveis(df_2024, termos_interesse)
variaveis_2025 = buscar_variaveis(df_2025, termos_interesse)

print(f"Variáveis candidatas em 2023-2024: {len(variaveis_2023)}")
print(f"Variáveis candidatas em 2024-2025: {len(variaveis_2024)}")
print(f"Variáveis candidatas em 2025-2026: {len(variaveis_2025)}")
```

### Interpretação da Busca por Variáveis Relevantes

Como a identificação foi realizada por correspondência textual nos nomes das colunas, algumas variáveis relevantes podem não ter sido capturadas, enquanto outras podem ter sido incluídas apenas por apresentarem termos semelhantes.

A definição do schema analítico será realizada posteriormente por meio da avaliação semântica das variáveis e de sua aderência às questões de negócio do Tech Challenge.

### 5.1 Identificação das Variáveis de Remuneração

A remuneração constitui uma dimensão relevante para a compreensão do mercado brasileiro de Dados e para a análise das diferenças entre perfis profissionais, senioridades, regiões e modelos de trabalho.

Nesta etapa, são identificadas as variáveis relacionadas à remuneração nas três edições da pesquisa, buscando posteriormente estabelecer correspondências semânticas entre os diferentes schemas.

```python
def localizar_colunas(df, termos):
    return [
        coluna for coluna in df.columns
        if any(termo in str(coluna).lower() for termo in termos)
    ]

termos_remuneracao = [
    "salario",
    "salarial",
    "remuneracao",
    "remuneração",
    "renda"
]

remuneracao_2023 = localizar_colunas(df_2023, termos_remuneracao)
remuneracao_2024 = localizar_colunas(df_2024, termos_remuneracao)
remuneracao_2025 = localizar_colunas(df_2025, termos_remuneracao)

print("=== 2023-2024 ===")
for coluna in remuneracao_2023:
    print(coluna)

print("\n=== 2024-2025 ===")
for coluna in remuneracao_2024:
    print(coluna)

print("\n=== 2025-2026 ===")
for coluna in remuneracao_2025:
    print(coluna)
```

### Interpretação das Variáveis de Remuneração

A análise identificou variáveis relacionadas à remuneração nas três edições da pesquisa.

A variável de **faixa salarial** apresenta correspondência semântica clara entre os períodos, sendo representada por `P2_h` em 2023-2024 e por `2.h_faixa_salarial` nas edições de 2024-2025 e 2025-2026.

Essa correspondência torna a faixa salarial uma candidata adequada para compor o schema harmonizado, permitindo análises relacionadas à remuneração ao longo dos diferentes períodos.

Também foram identificadas outras variáveis associadas a remuneração e salário. Entretanto, sua utilização dependerá da interpretação do contexto das respectivas perguntas, evitando combinar atributos que, apesar de possuírem termos semelhantes, representem conceitos distintos.
